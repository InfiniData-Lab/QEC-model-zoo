```python
import json
import re
from pathlib import Path
from typing import Dict, Union

import numpy as np
import pandas as pd


DATASET_NAMES = {
    1: "Dicarlo",
    2: "Google Surf d3",
    3: "Google Surf d5",
    4: "Google Rep d25",
    5: "Vandersypen",
    6: "Google Color d3",
    7: "Google Color d5",
    8: "Lukin d3",
    9: "Lukin d5",
}

OUTPUT_DIRECTORY = Path("experiments/results")
PLOT_OUTPUT_DIRECTORY = OUTPUT_DIRECTORY / "full_decoder_timing_plots"

TIMING_OUTPUT_FILE = OUTPUT_DIRECTORY / "full_decoder_timing_results.csv"
METRICS_OUTPUT_FILE = OUTPUT_DIRECTORY / "full_decoder_metric_results.csv"


def get_sorted_run_keys(run_results: dict) -> list[str]:
    """
    Sort run keys numerically so run_2 appears before run_10.
    """
    return sorted(
        run_results.keys(),
        key=lambda key: int(key.removeprefix("run_")),
    )


def compute_results(
    dataset_id: int,
    model_name: str,
    warmup_period: int = 5,
) -> Dict[str, Union[float, int]]:
    """
    Compute timing statistics and retrieve accuracy and LER.

    The first `warmup_period` runs are excluded from the timing
    calculations.

    Accuracy and LER are taken from the first run because they are
    assumed to be identical across runs.
    """
    results_path = result_mapping[dataset_id][model_name]

    with open(results_path, "r", encoding="utf-8") as file:
        results_file = json.load(file)

    run_results = results_file["metrics"]
    run_keys = get_sorted_run_keys(run_results)

    run_count_total = len(run_keys)
    shot_count = int(
        results_file["data_info"]["test"]["n_samples"]
    )

    if run_count_total == 0:
        raise ValueError(
            f"No runs found for dataset={dataset_id}, "
            f"model={model_name}."
        )

    if shot_count <= 0:
        raise ValueError(
            f"Invalid sample count for dataset={dataset_id}: "
            f"{shot_count}"
        )

    if warmup_period < 0:
        raise ValueError(
            f"Warmup period cannot be negative: {warmup_period}"
        )

    if warmup_period >= run_count_total:
        raise ValueError(
            f"Warmup period ({warmup_period}) must be smaller than "
            f"the number of runs ({run_count_total}) for "
            f"dataset={dataset_id}, model={model_name}."
        )

    # Accuracy and LER are assumed to be identical across runs.
    first_run = run_results[run_keys[0]]

    accuracy = float(first_run["accuracy"])
    ler = float(first_run["ler"])

    # Ignore warmup runs when calculating timing statistics.
    measured_run_keys = run_keys[warmup_period:]

    inference_times_per_shot_microseconds = []

    for run_key in measured_run_keys:
        inference_total_time_seconds = float(
            run_results[run_key]["inference_time"]
        )

        inference_time_per_shot_microseconds = (
            inference_total_time_seconds * 1e6 / shot_count
        )

        inference_times_per_shot_microseconds.append(
            inference_time_per_shot_microseconds
        )

    timing_values = np.asarray(
        inference_times_per_shot_microseconds,
        dtype=float,
    )

    mean_timing = float(np.mean(timing_values))

    # Sample standard deviation is generally appropriate when the
    # measured runs are considered a sample of repeated measurements.
    if len(timing_values) > 1:
        std_timing = float(np.std(timing_values, ddof=1))
    else:
        std_timing = 0.0

    return {
        "mean_us_per_shot": mean_timing,
        "std_us_per_shot": std_timing,
        "min_us_per_shot": float(np.min(timing_values)),
        "max_us_per_shot": float(np.max(timing_values)),
        "accuracy": accuracy,
        "ler": ler,
        "n_shots": shot_count,
        "n_measured_runs": len(measured_run_keys),
        "n_warmup_runs": warmup_period,
        "n_runs_total": run_count_total,
    }


def format_timing_cell(
    results: Dict[str, Union[float, int]],
) -> str:
    """
    Format timing information for one dataset/model table cell.
    """
    return (
        f"mean={results['mean_us_per_shot']:.2f} µs/shot; "
        f"std={results['std_us_per_shot']:.2f} µs/shot; "
    )


def format_metrics_cell(
    results: Dict[str, Union[float, int]],
) -> str:
    """
    Format accuracy and LER for one dataset/model table cell.
    """
    return (
        f"accuracy={results['accuracy']:.6f}; "
        f"LER={results['ler']:.6f}"
    )


def make_safe_filename(name: str) -> str:
    """
    Convert a dataset name into a filesystem-safe filename.
    """
    safe_name = re.sub(r"[^A-Za-z0-9_-]+", "_", name.strip())
    return safe_name.strip("_").lower()


# Preserve the model order based on the order in which each model
# first appears in result_mapping.
model_order = list(
    dict.fromkeys(
        model_name
        for model_mapping in result_mapping.values()
        for model_name in model_mapping.keys()
    )
)

timing_table = {}
metrics_table = {}

# Keep the numerical timing values separately for plotting.
timing_plot_results = {}

for dataset_id, model_mapping in result_mapping.items():
    dataset_name = DATASET_NAMES.get(
        dataset_id,
        f"Dataset {dataset_id}",
    )

    timing_table[dataset_name] = {}
    metrics_table[dataset_name] = {}

    for model_name in model_mapping.keys():
        results = compute_results(
            dataset_id=dataset_id,
            model_name=model_name,
            warmup_period=5,
        )

        timing_table[dataset_name][model_name] = (
            format_timing_cell(results)
        )

        metrics_table[dataset_name][model_name] = (
            format_metrics_cell(results)
        )

        print(
            f"Results for {model_name} on {dataset_name}: "
            f"{results}"
        )


# Create the two wide tables.
timing_df = pd.DataFrame.from_dict(
    timing_table,
    orient="index",
)

metrics_df = pd.DataFrame.from_dict(
    metrics_table,
    orient="index",
)

timing_df.index.name = "dataset"
timing_df.columns.name = "model"

metrics_df.index.name = "dataset"
metrics_df.columns.name = "model"


# Preserve the configured dataset order.
dataset_order = [
    dataset_name
    for dataset_id, dataset_name in DATASET_NAMES.items()
    if dataset_name in timing_df.index
]

# Add datasets not explicitly included in DATASET_NAMES afterward.
remaining_datasets = [
    dataset_name
    for dataset_name in timing_df.index
    if dataset_name not in dataset_order
]

ordered_datasets = dataset_order + remaining_datasets

timing_df = timing_df.reindex(index=ordered_datasets)
metrics_df = metrics_df.reindex(index=ordered_datasets)

# Preserve model order rather than sorting alphabetically.
timing_df = timing_df.reindex(columns=model_order)
metrics_df = metrics_df.reindex(columns=model_order)


# Create all output directories.
OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)


# Save the CSV tables.
timing_df.to_csv(
    TIMING_OUTPUT_FILE,
    encoding="utf-8",
)

metrics_df.to_csv(
    METRICS_OUTPUT_FILE,
    encoding="utf-8",
)


print(
    f"\nSaved timing table to: "
    f"{TIMING_OUTPUT_FILE.resolve()}"
)
print(timing_df.to_string())

print(
    f"\nSaved metrics table to: "
    f"{METRICS_OUTPUT_FILE.resolve()}"
)
print(metrics_df.to_string())
```

```python
# Check how LER=4 instead of 5
# Compare Lukin LER with r=5 versus direct inversion with r=4

import json
import pandas as pd
from IPython.display import display

rows = []

for dataset_id in [8, 9]:
    dataset_name = DATASET_NAMES[dataset_id]

    for model_name, results_path in result_mapping[dataset_id].items():
        with open(results_path, "r") as f:
            results = json.load(f)

        # Use the first run, consistent with the rest of the notebook
        metrics = results["metrics"]["run_1"]
        lep = float(metrics["lep"])
        saved_ler = float(metrics["ler"])

        # Direct inversion for both round counts
        ler_r5 = (1 - (1 - 2 * lep) ** (1 / 5)) / 2
        ler_r4 = (1 - (1 - 2 * lep) ** (1 / 4)) / 2

        rows.append({
            "Dataset": dataset_name,
            "Model": model_name,
            "LEP": lep,
            "Saved LER": saved_ler,
            "LER (r=5)": ler_r5,
            "LER (r=4)": ler_r4,
            "Difference (r4-r5)": ler_r4 - ler_r5,
        })

lukin_ler_df = pd.DataFrame(rows).set_index(["Dataset", "Model"])

display(
    lukin_ler_df.style.format({
        "LEP": "{:.6f}",
        "Saved LER": "{:.6f}",
        "LER (r=5)": "{:.6f}",
        "LER (r=4)": "{:.6f}",
        "Difference (r4-r5)": "{:+.6f}",
    })
)
```
